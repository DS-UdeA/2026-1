# **Hashing estático con colisiones y resolución mediante encadenamiento**

```python
class TablaHashEncadenamiento:
  def __init__(self, tamaño):
    self.m = tamaño
    self.tabla = [[] for _ in range(tamaño)]
  def hash(self, clave):
    return clave % self.m
  def insertar(self, clave):
    indice = self.hash(clave)
    self.tabla[indice].append(clave)
  def mostrar(self):
    for i, lista in enumerate(self.tabla):
      print(f"{i}: {lista}")

tabla = TablaHashEncadenamiento(7)
valores = [10, 22, 31, 4, 15, 28, 17]
for v in valores:
  tabla.insertar(v)

tabla.mostrar()
```
