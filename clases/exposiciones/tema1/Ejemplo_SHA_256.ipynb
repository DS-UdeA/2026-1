# **Ejemplo: SHA-256 en Python**

### **Código básico**

```python
import hashlib
# Mensaje (puede ser una contraseña)
mensaje = "miClave123"
# Convertir a bytes
mensaje_bytes = mensaje.encode()
# Aplicar SHA-256
hash_obj = hashlib.sha256(mensaje_bytes)
# Obtener el hash en formato hexadecimal
hash_hex = hash_obj.hexdigest()
print("Mensaje original:", mensaje)
print("Hash SHA-256:", hash_hex)
```

### **Ejemplo de sensibilidad**

```python
print(hashlib.sha256("hola".encode()).hexdigest())
print(hashlib.sha256("Hola".encode()).hexdigest())
```
