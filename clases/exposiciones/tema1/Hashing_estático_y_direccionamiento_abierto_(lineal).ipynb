# **Hashing estático con colisiones y resolución mediante direccionameinto abierto (lineal)**

```python
class TablaHashLineal:
  def __init__(self, tamaño):
    self.m = tamaño
    self.tabla = [None] * tamaño
  def hash(self, clave):
    return clave % self.m
  def insertar(self, clave):
    indice = self.hash(clave)
    while self.tabla[indice] is not None:
      indice = (indice + 1) % self.m
    self.tabla[indice] = clave
  def mostrar(self):
    print(self.tabla)

tabla = TablaHashLineal(7)
valores = [10, 22, 31, 4, 15, 28, 17]
for v in valores:
  tabla.insertar(v)

tabla.mostrar()
```
